產生文字、總結和分析內容只是您可以使用 Gemma 開放式模型完成的部分任務。本教學向您展示如何開始使用Hugging Face Transformers 執行Gemma，並使用文字和圖像輸入來產生文字內容。 Transformers Python library 提供了 API 用於存取預訓練的生成式 AI 模型，包括Gemma。有關更多信息，請參閱 [Transformers](https://huggingface.co/docs/transformers/en/index) 文件。
附註：Gemma 3以上機型支援影像及文字輸入。 Gemma 的早期版本僅支援文字輸入，某些變體除外，包括 [PaliGemma](https://ai.google.dev/gemma/docs/setup)。

## 從文本生成文本

使用文字提示 Gemma 模型以獲得文字回應是使用 Gemma 的最簡單方法，並且適用於幾乎所有 Gemma 變體。本節介紹如何使用 Hugging Face Transformers library 載入並設定 Gemma 模型以產生文字到文字。

### 負載模型

使用`torch` 和`transformers` 庫透過Gemma 建立模型執行`pipeline` 類別的實例。使用模型產生輸出或遵循指示時，請選擇指令調整 (IT) 模型，該模型的模型 ID 字串中通常包含`it`。使用 `pipeline` 對象，您可以指定要使用的 Gemma 變體、要執行的任務類型，特別是用於多模式生成的 `"any-to-any"`，如下列程式碼範例所示：

```
from transformers import pipeline

MODEL_ID = "google/gemma-4-E2B-it"

pipe = pipeline(
    task="any-to-any",
    model=MODEL_ID,
    device_map="auto",
    dtype="auto"
)
```

Gemma 僅支援少數 `task` 生成設定。有關可用 `task` 設定的更多信息，請參閱 Hugging Face 管道 [task()](https://huggingface.co/docs/transformers/main/en/main_classes/pipelines#transformers.pipeline.task) 文件。有關使用 Pipeline 類別的更多信息，請參閱 Hugging Face [Pipelines](https://huggingface.co/docs/transformers/main/en/main_classes/pipelines) 文件。

### 執行文字生成

在 `pipeline` 物件中載入並設定 Gemma 模型後，您可以將 prompt 傳送到模型。以下範例程式碼顯示了使用 `text` 參數的基本請求：

```
pipe(text="<|turn>user\nroses are red<turn|>\n<|turn>model\n")
```

### 使用 prompt 模板

當產生具有更複雜的 prompting 的內容時，請使用 prompt 範本來建立您的請求。 prompt 範本可讓您指定來自特定角色（例如`user` 或`model`）的輸入，並且是管理與Gemma 模型的多輪聊天互動的必要格式。以下範例程式碼顯示如何為Gemma建置prompt範本：

```
from transformers import GenerationConfig
config = GenerationConfig.from_pretrained(MODEL_ID)
config.max_new_tokens = 512
gen_kwargs = dict(generation_config=config)

messages = [
    {
        "role": "system",
        "content": [{"type": "text", "text": "You are a helpful assistant."}]
    },
    {
        "role": "user",
        "content": [{"type": "text", "text": "Roses are red..."}]
    },
]

pipe(messages, return_full_text=False, generate_kwargs=gen_kwargs)
```

<a name="vision"></a>## 從圖像資料生成文本

從 Gemma 3 開始，對於模型尺寸 4B 以上，您可以使用圖像資料作為 prompt 的一部分。本節介紹如何使用 Transformers library 載入和設定 Gemma 模型，以使用圖像資料和文字輸入產生文字輸出。

### 使用 prompt 模板

當產生具有更複雜的 prompting 的內容時，請使用 prompt 範本來建立您的請求。 prompt 範本可讓您指定來自特定角色（例如`user` 或`model`）的輸入，並且是管理與Gemma 模型的多輪聊天互動的必要格式。以下範例程式碼顯示如何為Gemma建置prompt範本：

```
from transformers import GenerationConfig
config = GenerationConfig.from_pretrained(MODEL_ID)
config.max_new_tokens = 512
gen_kwargs = dict(generation_config=config)

messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "url": "https://ai.google.dev/static/gemma/docs/images/thali-indian-plate.jpg"},
            {"type": "text", "text": "What is shown in this image?"},
        ]
    },
    {
        "role": "assistant",
        "content": [
            {"type": "text", "text": "This image shows"},
        ],
    },
]

pipe(text=messages, return_full_text=False, generate_kwargs=gen_kwargs)
```

您可以透過在 `content` 清單中新增 `"type": "image",` 條目，在 prompt 中新增多個圖像。
> 注意：請勿在prompt 模板的文字部分中使用`<|image|>`、`<start_of_image>` 或`<image_soft_token>` tokens 模板，因為此方法會產生冗餘tokens 和處理錯誤。

<a name="audio"></a>## 從音訊資料生成文字

透過 [Gemma 4](https://ai.google.dev/gemma/docs/gemma-4) 和 [Gemma 3n](https://ai.google.dev/gemma/docs/gemma-3n)，您可以使用音訊資料作為 prompt 的一部分。本節介紹如何使用 Transformers library 載入和設定 Gemma 模型以使用音訊資料和文字輸入產生文字輸出。

### 使用 prompt 模板

產生帶有音訊的內容時，請使用 prompt 模板來建立您的請求。 prompt 範本可讓您指定來自特定角色（例如`user` 或`model`）的輸入，並且是管理與Gemma 模型的多輪聊天互動的必要格式。以下範例程式碼展示如何使用音訊資料輸入為Gemma建置prompt範本：

```
from transformers import GenerationConfig
config = GenerationConfig.from_pretrained(MODEL_ID)
config.max_new_tokens = 512
gen_kwargs = dict(generation_config=config)

messages = [
    {
        "role": "user",
        "content": [
            {"type": "text", "text": "Transcribe the following speech segment in its original language. Follow these specific instructions for formatting the answer:\n* Only output the transcription, with no newlines.\n* When transcribing numbers, write the digits, i.e. write 1.7 and not one point seven, and write 3 instead of three."},
            {"type": "audio", "audio": "https://ai.google.dev/gemma/docs/audio/roses-are.wav"},
        ]
    }
]

pipe(text=messages, return_full_text=False, generate_kwargs=gen_kwargs)
```

您可以透過在 `content` 清單中新增 `"type": "audio",` 條目來在 prompt 中新增多個音訊檔案。
> 注意：請勿在prompt 範本的文字部分中使用`<|audio|>` 或`<audio_soft_token>` tokens 模板，因為此方法會產生冗餘tokens 和處理錯誤。
